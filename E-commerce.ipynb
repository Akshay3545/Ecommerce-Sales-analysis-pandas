```python
df = pd.read_csv(r'C:\Users\aksha\Downloads\Ecommerce_sales_datasets.csv')
```

```python
df.head()
```

# Top 10 States by Total Orders

```python
kuch_bhi = df.groupby(['State'], as_index=False)['Order'].sum().sort_values(by='Order', ascending=False).head(10)

sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data = kuch_bhi, x='State', y='Order')
```

# Top 5 States by Total Sales 

```python
kuch_bhi = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(5)

sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data = kuch_bhi, x='State', y='Amount')
```

# Product_Category-wise Orders

```python
sns.set(rc={'figure.figsize':(18,5)})
ax = sns.countplot(data = df, x='product_category')

for bars in ax.containers:
    ax.bar_label(bars)
           
```

```python
kuch_bhi = df.groupby(['product_category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(5)

sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data = kuch_bhi, x='product_category', y='Amount')
```
